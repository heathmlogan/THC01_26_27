# Importing Data
You'll want to import the data from the indicators in Eurostat but for demonstrating how to work with the data we will use the Excel "Dataviz.xls" from Brightspace.

This excel contains data on the Human Development Index, Material Footprint, and Ecological Consumption Fooprint for every most of the countries in the world in 2023 and is another way at looking at circular transition planning.



```python
import pandas as pd
# pandas is always imported as pd to make calling it easier when writing code

data = pd.read_excel("/content/dataviz_ws.xlsx", sheet_name = "dataviz")
# 'data' is the name of the dataframe we created from the excel
# "/content/dataviz_ws.xlsx" is the path to the file
# sheet_name = "dataviz" finds the sheet we want to pull data from

data
# data "prints"/displays the dataframe in the output below this box.
```

```python
# Scatter Plot

# import plotly express to load the functions we need for figures
import plotly.express as px

fig1 = px.scatter(data, x = "HDI ", y ="EFC (gha/pp)",
                  color="Country",
                  hover_data=['Country'],)
# Here we call the plot and data
# note this is how the boundaries for sustainability are often discussed by the UN

# You can add further details with the following
fig1.add_hline(y=1.5, line = dict(dash="dash"))
# 1.5 global hectares per person was the global biocapcity
fig1.add_vline(x=0.8, line = dict(dash="longdash"))
# 0.8 is the barrier for high human development
fig1.add_hrect(x0=0.8, x1=1, y0=0, y1=1.5,)

# You can update axis and labels with the following
fig1.update_yaxes(range=[0, 13])
fig1.update_xaxes(range=[0, 1])
fig1.layout.xaxis.title = "Human Development Index"
fig1.layout.yaxis.title = "Environmental Footprint Consumption (gha/pp)"


fig1.show()

fig2 = px.scatter(data, x = "HDI ", y ="MF (t/pp)", color="Country", hover_data=['Country'],)
# Here we call the plot and data
# note this is how the boundaries for sustainability are often discussed by the UN

# You can add further details with the following
fig2.add_hline(y=8, line = dict(dash="dash"))
# 8 tonnes per person per year is the global boundary for sustainability
fig2.add_vline(x=0.8, line = dict(dash="longdash"))
# 0.8 is the barrier for high human development
fig2.add_hrect(x0=0.8, x1=1, y0=0, y1=8,)

# You can update axis and labels with the following
fig2.update_yaxes(range=[0, 50])
fig2.update_xaxes(range=[0, 1])
fig2.layout.xaxis.title = "Human Development Index"
fig2.layout.yaxis.title = "Material Footprint (t/pp)"

fig2.show()

# Learn more about scatter plots here
# https://plotly.com/python/line-and-scatter/
```

# Interpretation
So in the two figures above we have two scatter plots, one showing the human development index (HDI) on the x-axis versus the Environmental Footprint for Consumption (EFC) on the y-axis, as well as the sufficiency boundary for environmentally sustainable development following the planetary Boundaries Framework and as part of the Sustainable Development Goals. And in the lower figure HDI is plotted against Material Footprint (MF) which is the boundary used by the Circular Gap Report we discussed on Tuesday which is used to differentiate which countries need to grow, build, or shift their economies towards a more circular future.  
Each graph is used by different actors to discuss how different countries should focus their regulatory efforts for a sustainable transition.

If you look at the position of counties by toggleing them on and off in the legend you may see that their position between figures shifts considerably, which can influence how diffent countires are prioritized. In addition, the focus is either on consumption (and wealth) or environmental impact, meaning we may prioritize countries and needs differently. So, what if we were to consider both aspects in tandem? How would this shift the perspective for each stakeholder?


```python
# Circle/Bubble Plot
import plotly.express as px
# import plotly

data_bubble = data.dropna()
# here we make a copy of our data and remove any rows with empty values.

fig_bubble = px.scatter(data_bubble, x = "HDI ", y ="EFC (gha/pp)",
                 size = "MF (t/pp)", hover_data=["Country"],
                 size_max=40, color = "MF (t/pp)")

# You can add further details with the following
fig_bubble.add_hline(y=1.5, line = dict(dash="dash"))
# 1.5 global hectares per person was the global biocapcity
fig_bubble.add_vline(x=0.8, line = dict(dash="longdash"))
# 0.8 is the barrier for high human development
fig_bubble.add_hrect(x0=0.8, x1=1, y0=0, y1=1.5,)

# You can update axis and labels with the following
fig_bubble.update_yaxes(range=[0, 13])
fig_bubble.update_xaxes(range=[0.0, 1.0])
fig_bubble.layout.xaxis.title = "Human Development Index"
fig_bubble.layout.yaxis.title = "Environmental Footprint Consumption (gha/pp)"
fig_bubble.show()

# learn more about bubble charts here
# https://plotly.com/python/bubble-charts/
```

```python
# Bar charts (vertical)
# pretty straight forward bar chart

data_bar_v = data.sort_values(by='MF (t/pp)', ascending=True)
# Here we sort the data frame for the way we want the figure to vizualize the data

fig_bar_v = px.bar(data_bar_v, x='Country', y='MF (t/pp)',
#change data to match the dataframe you create with pandas
# update x and y with the columns you want to compare
                   hover_data=['MF (t/pp)', 'pop_1000_persons'],
                   color='MF (t/pp)',
                   labels={'MF (t/pp)':'Material Footprint (tonnes/per person)'},
)

fig_bar_v.add_hline(y = 8,)
# adding our 'cutoff' for material consumption in tonnes per person

fig_bar_v.show()

# Want to see what else you can change? https://plotly.com/python/bar-charts/
```

```python
# Bar chart (horizontal - stacked)

# okay so we are going to sort the dataframe here so we can see the contribution of each material for the MF
df_bar_h = data_bar_v.melt(
    id_vars=['Country', 'HDI '],
    value_vars=[
        'MF Biomass (t/pp)',
        'MF Fossil fuels (t/pp)',
        'MF Metal ores (t/pp)',
        'MF Non-metallic minerals (t/pp)'
    ],
    var_name='Material Type',
    value_name='Material Footprint'
)

fig_bar_h = px.bar(df_bar_h, x='Material Footprint', y='Country',
    color='Material Type', orientation='h',
    hover_data=['HDI '], height =1000
)
# set up the horizontal bar chat as with the vertical barchat, just be sure to cahnge orientation

fig_bar_v.add_hline(y = 8,)
# adding our 'cutoff' for material consumption in tonnes per person

fig_bar_h.show()

# If you want to explore what else you can do https://plotly.com/python/horizontal-bar-charts/
```

# Heat Map

In the figure below we are using data from the EU Material Footprint Indicator. I wanted to see if any countries have reduced their per capita Material Consumption to below the 8-10 tonnes cutoff for sustainability across the years of reporting. To do this I used a heat map to look at the changes per country, but this could be done with the overall EU data as well.

One thing I changed was the colorscale, as I wanted anything below the value of 8 to be green and everything else to grow “warmer” so I divided 8/50 to see where to start my color scale. I split the rest the transitions equally, but you could actually use the averages of the dataset to better define the colorscale for the decision making context of the figure.   


```python
# Heat Map
import plotly.graph_objects as go

data_eu_mf = pd.read_excel("/content/dataviz_ws.xlsx", sheet_name = "EU_MF_tpc")
#reading an example of data from Eurostat

fig_heat = go.Figure(data = go.Heatmap(
        z=data_eu_mf.iloc[:, 1:].values, # using pandas functions to call all values after the 1st column
        x=data_eu_mf.columns[1:], # using pandas functions to call years for the columns
        y=data_eu_mf["Country"], # rows
        colorscale=[[0, 'green'], [.02, 'yellow'], [.5, 'red'],[1, 'purple']],
        # here I set the colorscale as described above so yellow starts to transition in after the cutoff
        colorbar_title="Material Footprint"
))

fig_heat.update_layout(
    title="Material Footprint per Capita by Country and Year",
    xaxis_title="Year",
    yaxis_title="Country",
    height = 1000,
)

fig_heat.show()

# want to learn about more things you can do with heatmaps?
# https://plotly.com/python/heatmaps/#basic-heatmap-with-plotlygraphobjects
```

```python
# Guage

fig_guage = go.Figure(go.Indicator(
    mode="gauge+number",
    value= data_eu_mf.loc[data_eu_mf["Country"] == "Belgium","2025"].iloc[0],
    title={'text': "Belgium Material Footprint (2025)"},
    gauge={
        'axis': {'range': [0, 20]},
        'bar': {'color': 'grey'},
        'steps': [
            {'range': [0, 8], 'color': 'green'},
            {'range': [8, 10], 'color': 'yellow'},
            {'range': [10, 15], 'color': 'orange'},
            {'range': [15, 20], 'color': 'red'},
        ],
        'threshold': {
            'line': {'color': 'Red',},
            'thickness': 1,
            'value': 8
        }
    }
))

fig_guage.show()
```

```python
# Line Chart

data_eu_mf_long = data_eu_mf.melt(
    id_vars="Country",
    var_name="Year",
    value_name="MF"
)

fig_line = px.line(data_eu_mf_long, x="Year", y="MF", color="Country",
                   title="Material Footprint per Capita by Country",
                   hover_data=["MF","Country"], height =1000
)
fig_line.add_hline(y = 8,)
# adding our 'cutoff' for material consumption in tonnes per person

fig_line.show()

# you can toggle the countries on and off individually or make a seperate graph

belgium = data_eu_mf_long[data_eu_mf_long["Country"] == "Belgium"]


fig_line_BE = px.line(belgium, x="Year", y="MF", color="Country",
                   title="Material Footprint per Capita by Country",
                   hover_data=["MF","Country"], height =400
)
fig_line_BE.add_hline(y = 8,)
# adding our 'cutoff' for material consumption in tonnes per person


fig_line_BE.show()
# to learn more about line graphs
# https://plotly.com/python/line-charts/
```
