 # Creating a figure

So first we need to import the package. The name of the package is “plotly.express” and it as almost always named “px” in code. This is just a python/plotly rule. I don’t know why, but just like learning any language there are some rules we choose to accept.

This means we have imported the files we need to create graphs and we call the package plotly.express “px” in our code from here on out, (which will save us the headache of typing plotly.express out every time). To create a figure we first assign an object, in this case we will call it fig_bar:

    fig_bar =[CODE HERE]

Next, we will need to add code to call the figure, this is where “px” comes in.

    [Object Name]=px.[Type of Graph]([DataFrame], x = [“Column A Name”], y = [“Column B Name”])

So for [Type of Graph], we want a “bar”, for [DataFrame] we want to call "EU_data", for [“Column A Name”] we call "Country”, and for we call MF(2023). This will result in the following code.

    fig_bar=px.bar(EU_data, x = “County”, y = “MF(2023)")

Then to show the graph we will type the name of the object and “.show()”.

    fig_bar.show()


```python
# bar chart (Requires 2 columns of data)
fig_bar=px.bar(EU_data, x = "Country", y = "MF(2023)",)

fig_bar.show()
```

```python
fig_bar=px.bar(EU_data, x = "Country", y = "MF(2023)", #data

              # The first line is the ony required line everything below is a nice to have addition

                title = "Material Footprint by EU Country in 2023",
               # Title of the figure

                hover_data = ["Country","MF(2023)"],
               # What you see when you hover your mouse on a bar in the figure

                color = "MF(2023)",
               # Assigns colors by country name try changing this to "MF(2023)" to see a difference

                color_discrete_sequence = px.colors.qualitative.Pastel,
               # Color scheme (try px.colors.qualitative.Plotly)

                template ="plotly",
               # Sets the general template for the figure (try "plotly_white" or "plotly_dark")

                labels = {"MF(2023)":'Material Footprint 2023'},
               # What the value in the assigned axis and hover_data is called

                height = 400,
               # 400 is a small figure 1000 is a tall figure

                width = 400,
               # width of the figure
 )


fig_bar.show()
```

## Single Year Call Outs
To show a single year we just call the bar chart as before but update the y column for the year we want to call.

    fig_bar_2020=px.bar(MF_yrs_data, x = "Country", y = "2020",)

and show the new graph

#Multiple Years
To show multiple years, we add all the years we want to call to the y axis

    fig_bar_yrs=px.bar(MF_yrs_data, x = "Country", y = ["2020","2021","2022"],)

And we need to update our figure layout to change the default display of the graph, which can be any of the following [“OPTION”]: 'stack', 'group', 'overlay', 'relative'.

    fig_bar_yrs.update_layout(barmode =[“OPTION”])

And in this case we choose to group

    fig_bar_yrs.update_layout(barmode = "group")

And because we have multiple data points on the y axis we update the label:

    fig_bar_yrs.layout.yaxis.title = "Material Footprint per year (t/pp)"

So the full code would look like:

    fig_bar_yrs=px.bar(MF_yrs_data,
    x = "Country",
    y = ["2020","2021","2022"],)

    fig_bar_yrs.update_layout(barmode = "group" )

    fig_bar_yrs.show()


```python
#show a single year
fig_bar_yrs=px.bar(MF_yrs_data, x = "Country", y = "2020",
                   labels = {"2020":"Material Footprint 2020"})

fig_bar_yrs.show()
```

```python
#show a multiple years
fig_bar_yrs=px.bar(MF_yrs_data,
                   x = "Country",
                   y = ["2020","2021","2022"],
                   )
fig_bar_yrs.update_layout(barmode = "group")

fig_bar_yrs.layout.yaxis.title = "Material Footprint per year (t/pp)"

fig_bar_yrs.show()
```

# Making Scatter and Line Plots: Transform the Data

So because our data looks like this:
|Country| 2020|2021|2022|
|--|--|--|--|
|DK| 1|2|3|
|NL|2|3|1|
|ESP|3|1|2|

We need to either invert our data or create a melt to plot our values against our years.

### Inverting data by Index

Inverting our data is just like when you copy something in excel and use the "paste special" function to switch the columns and rows. We use the ".setindex().T" function to do this.

So first we call our dataframe and then ".setindex("COLUMN NAME").T"

    df_i = df.setindex("COLUMN NAME").T

In our case:

    MF_yrs_data_i = MF_yrs_data.set_index("Country").T

This will then make our data look like:

| Country | DK | NL | ESP |
|------|---:|---:|----:|
| 2020 |  1 |  2 |   3 |
| 2021 |  2 |  3 |   1 |
| 2022 |  3 |  1 |   2 |

Now we want to re-name column "Country" to blank, reindex the table, and rename the new column with years:

    MF_yrs_data_i = MF_yrs_data.set_index("Country").T
    # turns the dataframe
    MF_yrs_data_i.columns.name = ""
    # resets the name of the first column to blank
    MF_yrs_data_i = MF_yrs_data_i.reset_index()
    # re indexes the df
    MF_yrs_data_i.rename(columns={"index": "Year"}, inplace=True)
    #names the first column "years"
    

And now if you print/display:  
    
    MF_yrs_data_i

You will see:

| Years | DK | NL | ESP |
|------|---:|---:|----:|
| 2020 |  1 |  2 |   3 |
| 2021 |  2 |  3 |   1 |
| 2022 |  3 |  1 |   2 |

### Inverting Data in Excel

You can always make a new sheet, save, re-upload, and create a new dataframe

### Inverting data via a melt

A melt create a different dataframe structure. Essentially it will combine all four columns into two columns, esentially elongating the dataframe. This is sometimes called long or melt depending on the code you are looking at.

To do this:

    MF_yrs_data_melt = MF_yrs_data.melt(
    id_vars="Country", # sets the indexing column
    # i.e. we will make 3 extra rows for each country
    var_name="Year", # This determines the name to the new column
    value_name="MF" # This asigns the value to the the new row)

And then when we print/display the dataframe

    MF_yrs_data_melt

It will look like this:

| Country | Year | MF |
|---------|------|---:|
| DK      | 2020 |  1 |
| NL      | 2020 |  2 |
| ESP     | 2020 |  3 |
| DK      | 2021 |  2 |
| NL      | 2021 |  3 |
| ESP     | 2021 |  1 |
| DK      | 2022 |  3 |
| NL      | 2022 |  1 |
| ESP     | 2022 |  2 |




```python
# Inverting the Dataframe
MF_yrs_data_i = MF_yrs_data.set_index("Country").T
# turns the dataframe
MF_yrs_data_i.columns.name = ""
# resets the name of the first column to blank
MF_yrs_data_i = MF_yrs_data_i.reset_index()
# re indexes the df
MF_yrs_data_i.rename(columns={"index": "Year"}, inplace=True)
# names the first column "years"


MF_yrs_data_i
```

```python
# Elongating/Melting the data frame

MF_yrs_data_melt = MF_yrs_data.melt(
    id_vars="Country", # sets the indexing column
    # i.e. we will make 3 extra rows for each country
    var_name="Year", # This determines the name to the new column
    value_name="MF", # This asigns the value to the the new row
    )

MF_yrs_data_melt
```

```python
# Line Chart with Inverted Data

fig_line_i= px.line(MF_yrs_data_i, x="Year", y=["DK","NL", "ESP"],
               )
fig_line_i.layout.yaxis.title = "Material Footprint (t/pp)"
fig_line_i.layout.legend.title = "Country"
fig_line_i.show()
```

```python
# Line Chart with Melted Data

fig_line_melt= px.line(MF_yrs_data_melt, x="Year", y=["MF"],
                  color = "Country",
               )
fig_line_melt.layout.yaxis.title = "Material Footprint (t/pp)"
fig_line_melt.show()
```

## Data Cleaning Rules to Keep in Mind

If you are comfortable with data cleaning and data science you don't have to follow these rules, but if you are new, these will make your life easier.

When in doubt look at how I have formated the different sheets in the Dataviz excel I provided as an example for both notebooks.

1. Make sure you have no data on the sheet except for the table

2. The first row should always be the table headers

3. Make sure each column has a header

4. If you make changes in your excel. Save and close the excel. If in colab delete and uplaod the new saved file. Then rerun the code block where you import the dataframe from excel.

5. Make sure there are no words or symbols in columns where you want values.

6. It is okay to leave a cell blank if there is no data, most the time (not always, but for all the graphs here) plotly will omit rows with blanks from figures, thus, only insert a zero in an empty cell if the value is truly zero.

# How to make a Map Figure

Notice in the new table we just imported you have the ISO standardized names for locations? This will allow you to make a country map of the material footprint for our ten countries.

So just like before we name our object:

    object = [CODE]

As we are making a map:

    fig_map = [CODE]

We will use the following code:

    fig_map = px.choropleth(EU_new_data,
                        locations="ISO",
                        color="MF (t/pp)",
                        hover_name="Country",)
    fig_map.show()

So we have named our figure, fig_map.

We call the plotly.express package by typing "px." then "choropleth" is the name for a map figure in Plotly.

EU_new_data is the DataFrame and the locations are found using the "ISO" column in out dataframe (table).

Then we color the countries by their material footprint using color="MF (t/pp)".

Finally when we place our mouse over a country we want it to tell us the name of the country not just the ISO abbreviation and MF (T/PP), so we write hover_name="Country" to call the name from the "Country" column.

Then we show the figure with:

    fig_map.show()


```python
fig_map = px.choropleth(EU_new_data,
                    locations = "ISO",
                    color = "MF (t/pp)",
                    hover_name = "Country",)
fig_map.show()
```

```python
#If I want to manually assign the clor scale I can do so by slightly altering the code
#Just like last time you can add and change details like title, label, hover data, template, etc.

fig_map = px.choropleth(EU_new_data, locations = "ISO",
                    title = "Material Footprint (tonnes/per person) by Country in 2023",
                    color = "MF (t/pp)",
                    hover_name = "Country",
                    color_continuous_scale = [
                      [0.0, "yellow"],
                          [0.5, "red"],
                           [1, "purple"]])
fig_map.show()

# Did you know when you hover over the map you can zoom in and out and move the map with your mouse?!
# If you Zoom in and use the Camera button in the output to take a screenshot
# This would let you fx. only focus on Europe in your data;
```

If you generate figures using code in your group work or MS Thesis its super helpful to include your notebook and the excel you use so an examiner or teacher can reproduce your figures. Ideally you also need README and TOML files but even providing a pre-run notebook like this and your data is sufficient. As I have done throughout, explaining each block of code and a brief interpretation is a good way of helping ensure even people less familiar with coding can follow along with what you have done. It's also a good idea to start citing your AI models, databases/datasets and literature in these notebooks.
